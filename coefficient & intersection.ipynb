```python
import numpy as np

x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]

# Find slope and intercept
m, c = np.polyfit(x, y, 1)

print("Coefficient (slope):", m)
print("Intercept:", c)
```
