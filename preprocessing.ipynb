```python
import pandas as pd

data = {
    'name':['Alice','Bob','Charlie','David','Emma','Frank','Grace',
            'Henry','Isabella','Jack','Karen','Liam','Mia','Noah','Olivia'],
    'gender':['Female','Male','Male','Male','Female','Male','Female',
              'Male','Female','Male','Female','Male','Female','Male','Female'],
    'age':[25,30,28,35,27,40,32,29,26,31,33,24,36,38,29],
    'department':['HR','IT','Finance','IT','HR','Sales','Finance',
                  'IT','Sales','HR','Finance','IT','Sales','Finance','HR'],
    'performance_score':[650,720,580,810,490,670,730,550,760,620,690,450,850,710,540]
}

df = pd.DataFrame(data)

df.to_csv('employee.csv', index=False)

print(df.head())
```

```python
import pandas as pd

df = pd.read_csv('employee.csv')

print(df.head())
print(df.info())
print(df.describe())
print("Shape:", df.shape)
```

```python
df.set_index('name', inplace=True)

print(df.head())
```

```python
high_perf = df[df['performance_score'] > 700]

print(high_perf)
```

```python
mid_perf = df[
    (df['performance_score'] > 500) &
    (df['performance_score'] < 700)
]

print(mid_perf)
```

```python
print(df.isnull().sum())
```

```python
df.fillna(df.select_dtypes(include='number').median(), inplace=True)
```

```python
lower = df['performance_score'].quantile(0.05)
upper = df['performance_score'].quantile(0.95)

print("Lower limit:", lower)
print("Upper limit:", upper)

df['performance_score'] = df['performance_score'].clip(lower, upper)
```

```python
outliers = df[
    (df['performance_score'] < lower) |
    (df['performance_score'] > upper)
]

print(outliers)
```

```python
df_encoded = pd.get_dummies(
    df,
    columns=['gender'],
    drop_first=True
)

print(df_encoded.head())
```
